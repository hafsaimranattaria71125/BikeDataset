```python
df.drop(columns=['instant','season','yr','mnth','holiday','weekday','workingday','casual','registered','atemp'])
```

```python
df['temp']=df['temp']*41
df['hum'] = df['hum'] * 100
df['windspeed'] = df['windspeed'] * 67
```

```python
grouped=(df.groupby('dteday',as_index=False).agg(
        cnt_sum=('cnt','sum'),
        weathersit_mod= ('weathersit',lambda x: x.mode().iloc[0] if not x.mode().empty else None ),
        temp_min=('temp','min'),
        temp_max=('temp','max'),
        temp_std=('temp','std'),
        temp_mean=('temp','mean'),

        hum_mean=('hum','mean'),
        hum_std=('hum','std'),
        windspeed_mean=('windspeed','mean'),
        windspeed_std=('windspeed','std'),
        hr_sum=('hr','sum')
))
grouped
```

```python
hr_sum=0
for i in range(24):
  hr_sum+=i
comp_df=grouped[grouped['hr_sum']==hr_sum]
fg=grouped[grouped['hr_sum']!=hr_sum]
comp_df
```

76 rows don't contain data for all day

train test split

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X=comp_df.drop(columns=['cnt_sum','dteday'],axis=1)
Y=comp_df['cnt_sum']
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)

residuals=y_test-y_pred
rmse=np.sqrt(mean_squared_error(y_test,y_pred))
r2=r2_score(y_test,y_pred)
rmse
```

```python
plt.scatter(y_pred,residuals)
plt.axhline(0, linestyle="--")
plt.xlabel("Fitted values (predicted rentals/day)")
plt.ylabel("Test residual (actual − fitted)")
plt.show()
```
